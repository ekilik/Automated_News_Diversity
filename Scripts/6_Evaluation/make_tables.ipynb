# Build results tables

## Setup

```python
import pandas as pd
import os
import numpy as np
import matplotlib.pyplot as plt

from docx import Document
from docx.shared import Pt
```

```python
# read excel
about_covid = pd.read_excel('data/All_results.xlsx', sheet_name='About covid',
                            header=0)
# rename columns
about_covid.columns = ['Method', 'Aantal Training', 'Aantal Validatie', 'Aantal Test', 'Accuracy', 'F1-macro', 'F1-weighted']
col_show_max = ['Method', 'Aantal Training', 'Aantal Validatie', 'Aantal Test', 'Accuracy', 'F1-macro', 'F1-weighted']
```

```python
#Create a Word document
doc = Document()
doc.add_heading('Table 1', level=1)  # Replace with your table title
# Add introductory text italics
intro_text = doc.add_paragraph()
intro_run = intro_text.add_run("Hoofdthema: Artikelen over Covid-19.")
intro_run.italic = True  # Make the text italic
intro_run.font.name = 'Times New Roman'  # Ensure APA formatting
intro_run.font.size = Pt(12)
```

```python
# Add a table with a custom structure
rows, cols = len(about_covid) + 2, len(about_covid.columns)  # Add 2 for headers
table = doc.add_table(rows=rows, cols=cols)
```

```python
for i, row in about_covid.iterrows():
    row_cells = table.rows[i + 2].cells
    for j, value in enumerate(row):
        row_cells[j].text = str(value)
```

```python
for row in table.rows:
    for cell in row.cells:
        for paragraph in cell.paragraphs:
            for run in paragraph.runs:
                run.font.name = 'Times New Roman'
                run.font.size = Pt(12)
```

```python
tbl = table._tbl  # Access the table's XML
for cell in tbl.iter("w:tcBorders"):
    for border in ["w:left", "w:right"]:
        element = cell.find(border)
        if element is not None:
            element.getparent().remove(element)
```

```python
output_path = 'apa_table.docx'
doc.save(output_path)
```
